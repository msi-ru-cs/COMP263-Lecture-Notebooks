## Filters

```python
horizontal_filter = np.array([
    [1, 2, 1],
    [0, 0, 0],
    [-1, -2, -1]])

vertical_filter = np.array([
    [1, 0, -1],
    [2, 0, -2],
    [1, 0, -1]])
```

```python
test_image = brick #astronaut_gray
test_image
```

```python
from numpy.matrixlib import test
# Apply filter
horizontal_edges = cv2.filter2D(test_image, -1, horizontal_filter)
vertical_edges = cv2.filter2D(test_image, -1, vertical_filter)
```

```python
# Display
plt.figure(figsize=(20, 12))
plt.subplot(1, 3, 1)
plt.imshow(test_image, cmap='gray')
plt.axis('off')
plt.title('Original Image')

plt.subplot(1, 3, 2)
plt.imshow(horizontal_edges, cmap='gray')
plt.axis('off')
plt.title('Horizontal Edges')

plt.subplot(1, 3, 3)
plt.imshow(vertical_edges, cmap='gray')
plt.axis('off')
plt.title('Vertical Edges')
```
